```python
CATALOG_NAME = "dbw_nenim_cloud_dba_dev"
SCHEMA_NAME = "northstar_gold"

STORAGE_ACCOUNT = "stnenimadlsdev01"

GOLD_ROOT = (
    f"abfss://gold@{STORAGE_ACCOUNT}.dfs.core.windows.net/northstar"
)

GOLD_TABLES = {
    "employee_summary": f"{GOLD_ROOT}/employee_summary",
    "department_summary": f"{GOLD_ROOT}/department_summary",
    "employer_summary": f"{GOLD_ROOT}/employer_summary",
    "state_summary": f"{GOLD_ROOT}/state_summary",
    "executive_kpis": f"{GOLD_ROOT}/executive_kpis",
    "eligibility_reconciliation": (
        f"{GOLD_ROOT}/eligibility_reconciliation"
    ),
    "eligibility_reconciliation_kpis": (
        f"{GOLD_ROOT}/eligibility_reconciliation_kpis"
    ),
    "data_quality_metrics": f"{GOLD_ROOT}/data_quality_metrics",
}

print(f"Catalog: {CATALOG_NAME}")
print(f"Schema:  {SCHEMA_NAME}")
```

```python
spark.sql(
    f"""
    CREATE SCHEMA IF NOT EXISTS
    `{CATALOG_NAME}`.`{SCHEMA_NAME}`
    """
)

print(
    f"Schema ready: "
    f"{CATALOG_NAME}.{SCHEMA_NAME}"
)
```

```python
for table_name, table_path in GOLD_TABLES.items():
    spark.sql(
        f"""
        CREATE TABLE IF NOT EXISTS
        `{CATALOG_NAME}`.`{SCHEMA_NAME}`.`{table_name}`
        USING DELTA
        LOCATION '{table_path}'
        """
    )

    print(
        f"Registered: "
        f"{CATALOG_NAME}.{SCHEMA_NAME}.{table_name}"
    )
```

```python
display(
    spark.sql(
        f"""
        SHOW TABLES IN
        `{CATALOG_NAME}`.`{SCHEMA_NAME}`
        """
    )
)
```

```python
for table_name in GOLD_TABLES:
    row_count = spark.table(
        f"{CATALOG_NAME}.{SCHEMA_NAME}.{table_name}"
    ).count()

    print(f"{table_name}: {row_count:,}")
```
